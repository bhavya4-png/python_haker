```python
def print_formatted(number):
    width = len(bin(number)) - 2  
    # your code goes here
    for i in range(1, number + 1):
        print(
            f"{i:{width}d} "
            f"{i:{width}o} "
            f"{i:{width}X} "
            f"{i:{width}b}"
        )

if __name__ == '__main__':
    n = int(input())
    print_formatted(n)
    
```
